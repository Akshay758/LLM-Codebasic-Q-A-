```python
from langchain_community.document_loaders import CSVLoader

loader = CSVLoader(file_path = "codebasics_faqs.csv", source_column = "prompt")
data=loader.load() 
```

```python
from langchain_huggingface import HuggingFaceEmbeddings

instructore_embeddings = HuggingFaceEmbeddings(
    model_name = "sentence-transformers/all-MiniLM-L6-v2"
)
```

```python
from langchain.vectorstores import Chroma
```

```python
vectordb=Chroma.from_documents(documents=data, embedding=instructore_embeddings)
```

```python
retriever = vectordb.as_retriever()
rdocs = retriever.get_relevant_documents("for how long is this course valid?")
rdocs
```

```python
from langchain.chains import create_retrieval_chain
from langchain.chains.combine_documents import create_stuff_documents_chain
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_template("""
Answer the question based only on the following context.

If the answer is not present in the context, say:
"I don't know based on the provided documents."

<context>
{context}
</context>

Question:
{input}

Answer:
""")

document_chain = create_stuff_documents_chain(
    llm,
    prompt
)

retrieval_chain = create_retrieval_chain(
    retriever,
    document_chain
)
```

```python
question = "Do you guys provide internship and also do you offer EMI payments?"

response = retrieval_chain.invoke({
    "input": question
})
```
